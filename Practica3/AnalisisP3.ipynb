### LED 1

```python
#Obtengo los datos del espectro del 1er LED
with open('Espectro-1.csv', mode='r') as archivo:
    ArchivoAbierto = csv.reader(archivo) #Usamos la biblioteca csv
    next(ArchivoAbierto,None)            #Me salto el primer renglón
    Lambda1, Counts1 = [], []
    for renglon in ArchivoAbierto:
        Lambda1.append(float(renglon[0]))
        Counts1.append(float(renglon[1]))
```

```python
#Obtenemos el valor máximo para reportar la lambda
count1_max = max(Counts1)
lambda1_max = Lambda1[np.argmax(Counts1)]

print('lambda max =',lambda1_max)
print('count max =',count1_max)
```

```python
#Consideramos su incertidumbre como el ancho del espectro (a la mitad del Count máximo)
print('(count max)/2 =',count1_max/2)

#Buscamos "a mano" en el Excel/csv los dos valores más cercanos a count_max#/2
count1_antes = 4630.83
lambda1_antes = 595.921

count1_despues = 4486.83
lambda1_despues = 583.929
```

```python
plt.plot(Lambda1,Counts1,'.',label=r'Mediciones')
plt.title(r'Espectro LED 1')
plt.xlabel(r'$\lambda$ [$nm$]'), plt.ylabel(r'counts')
plt.axvline(591.47330,c='r'), plt.axhline(9025.83,c='r') #Máximo
plt.axhline(count1_max/2,c='g')                          #Mitad
plt.axvline(595.921,c='purple'), plt.axhline(4630.83,c='purple') #Antes
plt.axvline(583.929,c='purple'), plt.axhline(4486.83,c='purple') #Después
plt.legend()
plt.grid()
plt.show()
```

```python
#Valor final para reportar
lambda1 = lambda1_max
inc_lambda1 = abs(lambda1_antes - lambda1_despues)

print(lambda1,'±',f"{inc_lambda1:.3f}")
```

Por lo tanto se trata de un "color" amarillo-naranja.

## Voltaje_diodo vs Frecuencia

```python
#Obtengo datos del V_diodo y de la Frecuencia
with open('Vdiodo-vs-Frec_v2.csv', mode='r') as archivo:
    ArchivoAbierto = csv.reader(archivo) #Usamos la biblioteca csv
    next(ArchivoAbierto,None)            #Me salto el primer renglón
    Vd, inc_Vd, Frec, inc_Frec = [], [], [], []
    for renglon in ArchivoAbierto:
        Vd.append(float(renglon[4]))
        inc_Vd.append(float(renglon[5]))
        Frec.append(float(renglon[6]))
        inc_Frec.append(float(renglon[7]))
```

```python
#plt.figure(figsize=(14,6))
plt.errorbar(Vd,Frec,xerr=inc_Vd,yerr=inc_Frec,fmt='.',label=r'Mediciones')
plt.title(r'Frecuencia de los LED vs. Voltaje del diodo')
plt.xlabel(r'$V_\text{diodo}$ [V]'), plt.ylabel(r'$\nu$ [$Hz$]')
plt.legend()
plt.grid()
plt.show()
```

```python
#Regresión lineal usando sklearn

Vd_reshape = np.array(Vd[1:]).reshape(-1,1) #Reescalamos V_d porque en X se necesita un arreglo 2D

model = LinearRegression() #Invocamos al modelo
model.fit(Vd_reshape, Frec[1:]) #Entrenamos al modelo

Frec_reg = model.predict(Vd_reshape) #Hacemos las predicciones (la regresión lineal para la Frecuencia como función de V_d)
```

**Nota: se ignoró el primer punto para hacer la regresión ya que es el único que está muy fuera de la tendencia.**

```python
plt.errorbar(Vd,Frec,xerr=inc_Vd,yerr=inc_Frec,fmt='.',label=r'Mediciones')
plt.plot(Vd[1:],Frec_reg,label='Ajuste')
plt.title(r'Frecuencia de los LED vs. Voltaje umbral')
plt.xlabel(r'$V_{U}$ [V]'), plt.ylabel(r'$\nu$ [$Hz$]')
plt.legend()
plt.grid()
plt.show()
```

```python
m = model.coef_[0]     #m: pendiente (e/h)
#b = model.intercept_   #b: ordenada al origen en A
razon_experimental = 1/m

#Calculo exacto de razon
e = constants.e
h = constants.h
razon_real = h / e

R2 = model.score(Vd_reshape, Frec[1:])

print('La razón REAL entre las constantes es de:', razon_real, 'en unidades del SI')
print('La razón EXPERIMENTAL entre las constantes es:',razon_experimental, 'en unidades del SI')
print('Tenemos un error relativo:',(razon_real-razon_experimental)/razon_experimental)
print('Y un coeficiente R^2 =', R2)
```
